<H3>Mutable Objects</H3>
<p>Some objects are mutable, meaning their values can change throughtout the executaion of a program. In practice, this means they maintain the same id even after a change in value. This is known as an 'in-place' change becase it doesn't require the creation of a new object to chang in value. Mutable objects include lists, dictionaries, and sets.</p>

<H3>Immutable Objects</H3>
<p>Immutable Objects on the other hand can not change in value during the execution ofa program. The only way to alter the value of a variable that points to an immutable object, is to create an entirely new object containing the altered value. In other words, immutable objects can not be changed 'in-place' Numbers, String, tuples are examples</p>

```python
l=[1,2,3]
print(l)
print(id(l))
l[1]=7
print(l)
print(id(l))
```

```python
a =10
b=10
print("a =",end=" ")
print(a)
print("id(a) =",end=" ")
print(id(a))
print("A and B having same  value b =",end=" ")
print(b)
print("A and B having same address id(b) =",end=" ")
print(id(b))
a=20 # address modified
print("Modified a =",end=" ")
print(a)
print("Modified id(a) =",end=" ")
print(id(a))
print("B having same  value b =",end=" ")
print(b)
print("B having same address id(b) =",end=" ")
print(id(b))
```

<b> Data Processing on File </b>

```python
file = open("dummy_file.txt");

contents = file.read();

type(contents)
```

```python
lines = contents.split("\n")

index =1

for line in lines:
    print(index, ")",line)
    print("\t Number of characters: ", len(line))
    words = line.split()
    print("\t Number of Words: ", len(words))
    index +=1

file.close()
```

<b> Using Exception with files</b>

```python
try:
    file = open("my_file.txt")
    file.read()
    file.close()
except Exception as e:
    print("Error > Some error happend in opening the file", e)
```
